```python
import numpy as np
import plotly.graph_objects as go

R = 2
r = 1
extent = R + r
samples = 50

thetas = np.linspace(0, 2 * np.pi, samples)
phis = np.linspace(0, 2 * np.pi, samples)

t, p = np.meshgrid(thetas, phis)

X = R * np.cos(t) + r * np.cos(t) * np.cos(p)
Y = R * np.sin(t) + r * np.sin(t) * np.cos(p)
Z = r * np.sin(p)

fig = go.Figure()

fig.add_trace(
    trace=go.Surface(
        x=X,
        y=Y,
        z=Z,
        lighting=dict(
            ambient=0.5,
            diffuse=0.8,
            specular=0.0,
            roughness=1,
            fresnel=1,
        ),
        colorscale=[[0, "royalblue"], [1, "royalblue"]],
        # opacity=0.5
    )
)

p_t = np.linspace(0, 2*np.pi, 10*samples)
p_p = np.zeros_like(p_t)

X = R * np.cos(p_t) + r * np.cos(p_t) * np.cos(p_p)
Y = R * np.sin(p_t) + r * np.sin(p_t) * np.cos(p_p)
Z = r * np.sin(p_p)

fig.add_trace(trace=go.Scatter3d(x=X, y=Y, z=Z, mode="lines"))
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math


def f(x, t):
    return t*x


def rk4(f, x):
    t = 0
    dt = 0.001
    positions = []
    while t < 10:
        positions.append((x, t))

        f1 = f(x, t)
        f2 = f(x + dt / 2 * f1, t + dt / 2)
        f3 = f(x + dt / 2 * f2, t + dt / 2)
        f4 = f(x + dt * f3, t + dt)

        x = x + dt / 6 * (f1 + 2 * f2 + 2 * f3 + f4)
        t += dt

    return positions

points = rk4(f, 0.001)
x, t = zip(*points)
plt.scatter(t, x)


```

```python
from PIL import Image
import numpy as np
from hashlib import sha256

h = sha256(b"Hello World!").digest()
h = np.array([int(c) for c in h]) % 3
h = h.reshape((8,4))

data = np.stack([np.zeros((8,4)),np.zeros((8,4)),np.zeros((8,4))])

data[0,h == 0] = 255
data[1,h == 1] = 255
data[2,h == 2] = 255

Image.fromarray(data.transpose((1,2,0)).astype(np.uint8)).resize((800, 400), Image.NEAREST)
```
